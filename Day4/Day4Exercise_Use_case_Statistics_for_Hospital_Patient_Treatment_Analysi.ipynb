```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    "Patient_ID": range(1,101),
    "Recovered": np.random.choice(["Yes","No"],100),
    "Treatment_Type": np.random.choice(["Medication","Therapy","Surgery"],100),
    "Disease_Type": np.random.choice(["Cardiac","Diabetes","Respiratory"],100),
    "Insurance": np.random.choice(["Yes","No"],100),
    "Treatment_Cost": np.random.normal(50000,10000,100)
}

df = pd.DataFrame(data)
df.head()
```

```python
# Probability that a patient recovers

recovery_prob = (df["Recovered"] == "Yes").mean()

print("Probability of Recovery:", recovery_prob)
```

```python
# Recovery probability by treatment type

pd.crosstab(
    df["Treatment_Type"],
    df["Recovered"],
    normalize="index"
)
```

```python
# Event A = Surgery
# Event B = Medication

surgery_patients = df[df["Treatment_Type"] == "Surgery"]
medication_patients = df[df["Treatment_Type"] == "Medication"]

# Since one patient can have only one treatment type,
# intersection is always zero

intersection = len(
    df[
        (df["Treatment_Type"] == "Surgery") &
        (df["Treatment_Type"] == "Medication")
    ]
)

print("Number of patients in A ∩ B:", intersection)

if intersection == 0:
    print("\nInterpretation:")
    print("P(A ∩ B) = 0")
    print("A patient cannot receive Surgery and Medication simultaneously.")
    print("Therefore, the events are mutually exclusive.")
```

```python
# H0: Treatment type does not affect recovery
# H1: Treatment type affects recovery

print("H0: Treatment type does not affect recovery")
print("H1: Treatment type affects recovery")
```

```python
# One-Tail vs Two-Tail Test

print("Scenario:")
print("Hospital management believes Surgery has a higher recovery rate than Medication.\n")

print("One-Tail Test")
print("H0: Surgery recovery rate <= Medication recovery rate")
print("H1: Surgery recovery rate > Medication recovery rate")

print("\nTwo-Tail Test")
print("H0: Surgery recovery rate = Medication recovery rate")
print("H1: Surgery recovery rate != Medication recovery rate")


# Recovery rates by treatment type

recovery_rates = pd.crosstab(
    df["Treatment_Type"],
    df["Recovered"],
    normalize="index"
)

print(recovery_rates)

surgery_rate = recovery_rates.loc["Surgery", "Yes"]
medication_rate = recovery_rates.loc["Medication", "Yes"]

print("\nSurgery Recovery Rate:", round(surgery_rate, 3))
print("Medication Recovery Rate:", round(medication_rate, 3))

if surgery_rate > medication_rate:
    print("\nObservation:")
    print("Surgery shows a higher recovery rate than Medication.")
    print("A One-Tail test would investigate whether this difference is statistically significant.")
else:
    print("\nObservation:")
    print("Surgery does not show a higher recovery rate than Medication.")

    print("One-Tail Test:")
print("Used when we want to test whether Surgery performs better than Medication.")

print("\nTwo-Tail Test:")
print("Used when we want to test whether any difference exists between treatment methods.")
```

```python
from scipy.stats import chi2_contingency

# Contingency table
table = pd.crosstab(
    df["Treatment_Type"],
    df["Recovered"]
)

# Chi-Square Test
chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("p-value:", p)
print("Degrees of Freedom:", dof)

if p < 0.05:
    print("Reject H0")
    print("Treatment type significantly affects recovery")
else:
    print("Fail to Reject H0")
    print("No significant relationship found")
```

```python
from scipy.stats import f_oneway

# Treatment cost by treatment type
medication = df[df["Treatment_Type"]=="Medication"]["Treatment_Cost"]

therapy = df[df["Treatment_Type"]=="Therapy"]["Treatment_Cost"]

surgery = df[df["Treatment_Type"]=="Surgery"]["Treatment_Cost"]

# ANOVA Test
f_stat, p_value = f_oneway(
    medication,
    therapy,
    surgery
)

print("F Statistic:", f_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
    print("Treatment costs differ significantly")
else:
    print("Fail to Reject H0")
    print("No significant cost difference")
```
