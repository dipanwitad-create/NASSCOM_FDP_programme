```python
# --------------------------------------------------
# Task 2: Synthetic Data Generation
# --------------------------------------------------

dates = pd.date_range('2023-01-01', periods=100)

df_sales = pd.DataFrame({
    'Date': dates,
    'Region': np.random.choice(
        ['East', 'West', 'North', 'South'], 100),
    'Product': np.random.choice(
        ['Laptop', 'Mouse', 'Keyboard', 'Monitor'], 100),
    'Sales': np.random.randint(100, 1001, 100),
    'Quantity': np.random.randint(1, 11, 100)
})

# Introduce NaN values
sales_nan_idx = np.random.choice(df_sales.index, 5, replace=False)
qty_nan_idx = np.random.choice(df_sales.index, 3, replace=False)

df_sales.loc[sales_nan_idx, 'Sales'] = np.nan
df_sales.loc[qty_nan_idx, 'Quantity'] = np.nan

print("\nFirst 5 rows:")
print(df_sales.head())
```

```python
# --------------------------------------------------
# Task 3: Indexing and Slicing
# --------------------------------------------------

# 3.1 Region and Sales columns
df3_1 = df_sales[['Region', 'Sales']]

# 3.2 Product = Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# 3.3 East and Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

# 3.4 First Monitor Sales value
val3_4 = df_sales.loc[
    df_sales['Product'] == 'Monitor',
    'Sales'
].iloc[0]

print("\nFirst Monitor Sales:", val3_4)
```

```python
# --------------------------------------------------
# Task 3: Indexing and Slicing
# --------------------------------------------------

# 3.1 Region and Sales columns
df3_1 = df_sales[['Region', 'Sales']]

# 3.2 Product = Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']

# 3.3 East and Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]

# 3.4 First Monitor Sales value
val3_4 = df_sales.loc[
    df_sales['Product'] == 'Monitor',
    'Sales'
].iloc[0]

print("\nFirst Monitor Sales:", val3_4)
```

```python
# --------------------------------------------------
# Task 5: Data Manipulation
# --------------------------------------------------

# 5.1 Total Revenue
df_sales['Total_Revenue'] = (
    df_sales['Sales'] * df_sales['Quantity']
)

# 5.2 Sort
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

# 5.3 Group by Region
df5_3_grouped_region = df_sales.groupby('Region')[
    ['Sales', 'Quantity']
].sum()

print("\nSales & Quantity by Region:")
print(df5_3_grouped_region)

# 5.4 Average Sales by Product
df5_4_grouped_product = df_sales.groupby(
    'Product'
)['Sales'].mean()

print("\nAverage Sales by Product:")
print(df5_4_grouped_product)
```
